```python
import matplotlib.pyplot as plt
import numpy as np

# Constants for capabilities and colors
ShaclSimpleNotOpt = 'SHACL'
NoshaclSimple = 'No SHACL'
ScSimple = 'Smart Contract'

ColorShaclSimpleNotOpt = '#92C5F9'
ColorNoshaclSimple = '#AFDC8F'
ColorScSimple = '#F8AE54'

# Data for workloads
workloads = ['5 TPS', '20 TPS']

# Create data arrays organized by capability
shacl_data = [4.883716117, 3.832625747]      # SHACL
noshacl_data = [5.134218417, 3.665766961]    # No SHACL
sc_data = [0.17, 0.17]                        # Smart Contract

# Create figure
fig, ax = plt.subplots(figsize=(10, 6))

# Add grid lines for better readability (with zorder=0 to place behind bars)
ax.grid(axis='y', linestyle='--', alpha=0.7, zorder=0)

# Bar width and positions
bar_width = 0.25
index = np.arange(len(workloads))

# Create bars for each capability (with zorder=2 to place in front of grid)
bars = []
bars.append(ax.bar(index - bar_width, shacl_data, bar_width, color=ColorShaclSimpleNotOpt, 
                   edgecolor='black', label=ShaclSimpleNotOpt, zorder=2))
bars.append(ax.bar(index, noshacl_data, bar_width, color=ColorNoshaclSimple, 
                   edgecolor='black', label=NoshaclSimple, zorder=2))
bars.append(ax.bar(index + bar_width, sc_data, bar_width, color=ColorScSimple, 
                   edgecolor='black', label=ScSimple, zorder=2))

# Set labels and title
ax.set_title('Throughput by Workload and Capability', fontsize=16, fontweight='bold')
ax.set_xlabel('Workload', fontsize=16, fontweight='bold')
ax.set_ylabel('Throughput [TPS]', fontsize=16, fontweight='bold')

# Adjust x-tick positions and labels
ax.set_xticks(index)
ax.set_xticklabels(workloads, fontsize=14, fontweight='bold')

# Add legend
ax.legend(fontsize=12)

# Layout adjustment
plt.tight_layout()

# Save figure
plt.savefig("chart-tps.pdf", format='pdf')

# Show the plot
plt.show()
```
